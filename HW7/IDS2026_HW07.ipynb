**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
# Due to the fact that some of our datasets are in json we first need to convert them to cvs.
import json
import pandas as pd

# The three years our dataset will be working with
aastad = [2022, 2023, 2024]

for aasta in aastad:

    fail = open(f"nvdcve-2.0-{aasta}.json", encoding="utf-8")
    andmed = json.load(fail)
    fail.close()

    read = []

    # Check all of the vunerabilities
    for kirje in andmed["vulnerabilities"]:

        cve = kirje["cve"]
        cve_id = cve["id"]
        avaldatud = cve["published"]

        kirjeldus = ""

        for tekst in cve["descriptions"]:
            if tekst["lang"] == "en":
                kirjeldus = tekst["value"]

        skoor = None
        raskusaste = None
        rundevektor = None
        rundekeerukus = None
        kasutaja_tegevus = None

        if "cvssMetricV31" in cve.get("metrics", {}):
            cvss = cve["metrics"]["cvssMetricV31"][0]["cvssData"]

            skoor = cvss["baseScore"]
            raskusaste = cvss["baseSeverity"]
            rundevektor = cvss["attackVector"]
            rundekeerukus = cvss["attackComplexity"]
            kasutaja_tegevus = cvss["userInteraction"]

        read.append({
            "cve_id": cve_id,
            "published": avaldatud,
            "description": kirjeldus,
            "cvss_score": skoor,
            "severity": raskusaste,
            "attack_vector": rundevektor,
            "attack_complexity": rundekeerukus,
            "user_interaction": kasutaja_tegevus
        })

    tabel = pd.DataFrame(read)
    tabel.to_csv(f"nvd_{aasta}.csv", index=False)
    print(aasta, "valmis!", len(tabel), "rida")
```

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
## TODO

import matplotlib.pyplot as plt

# combining the three NVD datasets
nvd = pd.concat([nvd_2022, nvd_2023, nvd_2024], ignore_index=True)

# adding KEV membership, 1 = in KEV, 0 = not in KEV
nvd["kev"] = nvd["cve_id"].isin(kev["cveID"]).astype(int)

# number of vulnerabilities by KEV membership
nvd["kev"].value_counts().sort_index().plot(kind="bar")

plt.title("Vulnerabilities by KEV Membership")
plt.xlabel("KEV membership (0 = No, 1 = Yes)")
plt.ylabel("Number of vulnerabilities")
plt.xticks(rotation=0)
plt.show()

# CVSS score distribution
nvd[nvd["kev"] == 0]["cvss_score"].hist(bins=20, alpha=0.5, density=True, label="Not in KEV")
nvd[nvd["kev"] == 1]["cvss_score"].hist(bins=20, alpha=0.5, density=True, label="In KEV")

plt.title("CVSS Score Distribution")
plt.xlabel("CVSS Score")
plt.ylabel("Density")
plt.legend()
plt.show()

# KEV percentage by attack vector
rates = nvd.groupby("attack_vector")["kev"].mean() * 100

rates.plot(kind="bar")

plt.title("KEV Membership by Attack Vector")
plt.xlabel("Attack Vector")
plt.ylabel("Percentage in KEV (%)")
plt.xticks(rotation=0)
plt.show()
```

**Answer:** We created three visualizations to explore the relationships between vulnerability characteristics and CISA KEV membership.

The first graph shows a strong class imbalance. The vast majority of vulnerabilities are not included in the KEV catalog, while only a very small proportion are included. This suggests that accuracy alone may not be a suitable metric for evaluating our classification models.

The second graph shows that KEV-listed vulnerabilities tend to have higher CVSS scores, with particularly high concentrations around 7.5, 8.8, and 9.8–10. In contrast, vulnerabilities not included in KEV have a wider distribution, including more medium-severity scores.

The third graph shows that KEV inclusion rates are low across all attack vectors. Network-based vulnerabilities have the highest inclusion rate at ~ 0.52%, followed by local at ~ 0.45% and adjacent-network vulnerabilities at ~ 0.36%. Physical vulnerabilities have no KEV matches in our dataset.

Overall, the visualizations suggest that severity scores and attack vectors may help distinguish KEV-listed vulnerabilities from non-listed ones. However, the substantial class imbalance will need to be considered during model development and evaluation.

**d. Check for data quality**

```python
## TODO

# missing values
print("NVD missing values:")
print(nvd.isna().sum())

print("\nCISA KEV missing values:")
print(kev.isna().sum())

# duplicate CVE identifiers
print("\nDuplicate CVE IDs in NVD:")
print(nvd["cve_id"].duplicated().sum())

print("\nDuplicate CVE IDs in KEV:")
print(kev["cveID"].duplicated().sum())

# check impossible CVSS scores
invalid_scores = nvd[
    (nvd["cvss_score"] < 0) |
    (nvd["cvss_score"] > 10)
]

print("\nInvalid CVSS scores:")
print(len(invalid_scores))

# publication dates
nvd["published"] = pd.to_datetime(
    nvd["published"], errors="coerce"
)

print("\nInvalid publication dates:")
print(nvd["published"].isna().sum())

# possible CVSS score outliers
print("\nCVSS score statistics:")
print(nvd["cvss_score"].describe())

# category values
print("\nSeverity categories:")
print(nvd["severity"].value_counts(dropna=False))

print("\nAttack vector categories:")
print(nvd["attack_vector"].value_counts(dropna=False))
```

We found no duplicate CVE identifiers in either dataset, no invalid publication dates, and no CVSS scores outside the valid range of 0–10. The CVSS scores range from 0 to 10, with a mean of approximately 6.79. Most vulnerabilities have medium or high severity, and NETWORK is the most common attack vector.

For further analysis, we plan to investigate the missing CVSS information and determine how to handle incomplete records. We will keep valid extreme scores, as they represent legitimate severity levels, and exclude invalid or duplicate records if any are identified later.

Overall, the datasets appear suitable for our analysis, although missing technical information and the strong imbalance between KEV-listed and non-listed vulnerabilities will need to be considered during modeling.
